```python
pearson_stat, pearson_p = stats.pearsonr(df['alcohol'], df['quality'])

print(f"Pearson Correlation Coefficient: {pearson_stat:.4f}")
print(f"P-value: {pearson_p:.4e}")
```

```python
spearman_stat, spearman_p = stats.spearmanr(df['alcohol'], df['quality'])

print(f"Spearman Rank Correlation Coefficient: {spearman_stat:.4f}")
print(f"P-value: {spearman_p:.4e}")
```

```python
low_quality_alc = df[df['quality'] <= 5]['alcohol']
high_quality_alc = df[df['quality'] >= 7]['alcohol']

ttest_stat, ttest_p = stats.ttest_ind(high_quality_alc, low_quality_alc, equal_var=False)

mwu_stat, mwu_p = stats.mannwhitneyu(high_quality_alc, low_quality_alc)

print(f"Welch's t-test: t-statistic = {ttest_stat:.4f}, p-value = {ttest_p:.4e}")
print(f"Mann-Whitney U test: statistic = {mwu_stat:.4f}, p-value = {mwu_p:.4e}")
```

```python
np.random.seed(42)
sample_alcohol = np.random.choice(df['alcohol'], size=500, replace=False)

shapiro_stat, shapiro_p = stats.shapiro(sample_alcohol)

print(f"Shapiro-Wilk Test Statistic: {shapiro_stat:.4f}")
print(f"P-value: {shapiro_p:.4e}")
```

```python
mean_alcohol = df['alcohol'].mean()
sem_alcohol = stats.sem(df['alcohol'])

ci_lower, ci_upper = stats.t.interval(
    confidence=0.95,
    df=len(df['alcohol']) - 1,
    loc=mean_alcohol,
    scale=sem_alcohol)

print(f"Sample Mean Alcohol: {mean_alcohol:.4f}%")
print(f"95% Confidence Interval: [{ci_lower:.4f}%, {ci_upper:.4f}%]")
```

1. Intercept (const = −15.6777, p=0.024)
Sign & Meaning: Negative.
Interpretation: Mathematically, it represents the theoretical baseline quality score when all predictor variables are set to zero. While it has no direct real-world physical meaning (since wine cannot have zero alcohol or density), it serves as the essential vertical anchor that positions the regression plane correctly.

2. Alcohol (alcohol = +0.3863, p<0.001)
Sign & Meaning: Positive (+).
Interpretation: For every 1% increase in alcohol volume, the wine quality score increases by 0.39 points, holding all other chemical properties constant. This aligns with EDA findings, confirming alcohol as the strongest positive predictor of expert quality ratings.

3. Volatile Acidity (volatile acidity = −1.9037, p<0.001)
Sign & Meaning: Negative (−).
Interpretation: Higher volatile acidity severely penalizes wine ratings. Each unit increase drops the quality score by 1.90 points. This reflects domain reality: high acetic acid causes unpleasant, vinegary off-flavors that degrade quality.

4. Chlorides (chlorides = −1.3377, p=0.018)
Sign & Meaning: Negative (−).
Interpretation: Increased salt content negatively affects quality scores. Higher chloride levels lead to a harsher or saltier taste profile, reducing consumer and expert appeal.

5. Total Sulfur Dioxide (total sulfur dioxide = +0.0010, p=0.002)
Sign & Meaning: Positive (+).
Interpretation: While bivariate correlation showed a slight negative trend, the partial coefficient in the multi-variable model is slightly positive (≈0.001 per ppm). When controlling for volatile acidity and sulfites, adequate sulfur dioxide properly preserves the wine against oxidation without introducing harsh chemical flaws.

6. Density (density = +18.0070, p=0.009)
Sign & Meaning: Positive (+).
Interpretation: Although density has a strong negative bivariate correlation with quality (r=−0.34), its coefficient flips to positive in this OLS model due to collinearity with alcohol (since ethanol is significantly lighter than water). Controlling for alcohol allows density to capture secondary balancing effects in the sugar-water-alcohol matrix.

```python
X = sm.add_constant(df[features])
y = df['quality']
model = sm.OLS(y, X).fit()

p_values = model.pvalues
conf_intervals = model.conf_int(alpha=0.05)
conf_intervals.columns = ['Lower CI (2.5%)', 'Upper CI (97.5%)']

summary_df = pd.DataFrame({
    'Coefficient': model.params,
    'p-value': p_values,
    'Lower CI': conf_intervals['Lower CI (2.5%)'],
    'Upper CI': conf_intervals['Upper CI (97.5%)']
})
print(summary_df)
```

```python
print(f"R-squared: {model.rsquared:.4f}")
print(f"Adjusted R-squared: {model.rsquared_adj:.4f}")
print(f"F-statistic: {model.fvalue:.4f}")
print(f"F-statistic p-value: {model.f_pvalue:.4e}")
```

model's limitations
1. Ratings are Whole Numbers: Wine quality is graded on a scale from 3 to 9, but the math formula treats them like smooth decimals. In reality, moving from a 5 to a 6 isn't always the exact same "distance" as moving from a 7 to an 8.

2. We Missed Other Factors: Our model only explains about 26.5% of what makes a wine good. The other 73.5% depends on things we didn't include, like the grape's region, weather, vintage year, or other chemical compounds.

3. Uneven Errors: The model's prediction mistakes (errors) aren't spread out evenly—it tends to make bigger mistakes on certain types of wine than others.

4. Straight-Line Assumption: The model assumes that adding more of a chemical always has a steady, straight-line effect on quality, whereas real-life chemistry often has tipping points or limits.terations.

```python

df = pd.read_csv('wine_quality_cleaned.csv')

y = df['quality']

features = [
    'fixed acidity', 'volatile acidity', 'citric acid', 'residual sugar', 
    'chlorides', 'free sulfur dioxide', 'total sulfur dioxide', 'density', 
    'pH', 'sulphates', 'alcohol'
]
X = df[features]

print("Target variable defined successfully!")
print(f"Target variable (y) shape: {y.shape}")
print(f"Feature matrix (X) shape: {X.shape}")
```

```python
X = df[features]
y = df['quality']

print(f"Successfully selected {len(features)} predictor variables for machine learning regression.")
print(X.info())
```

```python
from sklearn.model_selection import train_test_split


X_train, X_test, y_train, y_test = train_test_split(
    X, y, 
    test_size=0.20, 
    random_state=42
)

print(f"Training set shape: X_train = {X_train.shape}, y_train = {y_train.shape}")
print(f"Testing set shape:  X_test = {X_test.shape},  y_test = {y_test.shape}")
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
```

```python
lr_model = LinearRegression()
dt_model = DecisionTreeRegressor(random_state=42)
rf_model = RandomForestRegressor(random_state=42)

lr_model.fit(X_train, y_train)
dt_model.fit(X_train, y_train)
rf_model.fit(X_train, y_train)

print("Models successfully trained:")
print("- Linear Regression")
print("- Decision Tree Regressor")
print("- Random Forest Regressor")
```

```python
y_pred_lr = lr_model.predict(X_test)
y_pred_dt = dt_model.predict(X_test)
y_pred_rf = rf_model.predict(X_test)

print("Test set predictions generated successfully!")
print("Linear Regression predictions sample:", y_pred_lr[:5].round(2))
print("Decision Tree predictions sample:    ", y_pred_dt[:5])
print("Random Forest predictions sample:  ", y_pred_rf[:5].round(2))
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

model_predictions = {
    'Linear Regression': y_pred_lr,
    'Decision Tree': y_pred_dt,
    'Random Forest': y_pred_rf
}

evaluation_results = []

for name, y_pred in model_predictions.items():
    mae = mean_absolute_error(y_test, y_pred)
    mse = mean_squared_error(y_test, y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_test, y_pred)
    
    evaluation_results.append({
        'Model': name,
        'MAE': mae,
        'MSE': mse,
        'RMSE': rmse,
        'R^2 Score': r2
    })

# Compile into a comparative DataFrame
performance_df = pd.DataFrame(evaluation_results)
print(performance_df.to_string(index=False))
```

```python
sns.set_theme(style="whitegrid")

fig, axes = plt.subplots(1, 3, figsize=(18, 5), sharey=True)

models_dict = {
    'Linear Regression': y_pred_lr,
    'Decision Tree': y_pred_dt,
    'Random Forest': y_pred_rf
}

for ax, (name, y_pred) in zip(axes, models_dict.items()):
    # Scatter plot of Actual vs. Predicted values
    sns.scatterplot(x=y_test, y=y_pred, ax=ax, alpha=0.4, color='royalblue', edgecolor=None)
    
    # Draw the ideal 45-degree reference line (y = x) representing perfect predictions
    min_val = min(y_test.min(), y_pred.min())
    max_val = max(y_test.max(), y_pred.max())
    ax.plot([min_val, max_val], [min_val, max_val], color='crimson', linestyle='--', lw=2, label='Ideal Fit (y = x)')
    
    # Configure titles and labels
    ax.set_title(f'{name}', fontsize=12, fontweight='bold')
    ax.set_xlabel('Actual Quality Score', fontsize=10)
    if ax == axes[0]:
        ax.set_ylabel('Predicted Quality Score', fontsize=10)
    else:
        ax.set_ylabel('')
        
    ax.legend(loc='upper left')

plt.suptitle('Model Performance: Actual vs. Predicted Wine Quality', fontsize=14, fontweight='bold', y=1.03)
plt.tight_layout()
plt.show()
```

using a simple and justified selection criterion:
Highest $R^2$ Accuracy: Random Forest achieves the highest $R^2$ score (~50%), explaining significantly more variance in wine quality than Linear Regression (~27%) or Decision Tree (~32–35%). 
Lowest Error: It produces the lowest error rates (RMSE and MAE), meaning its predicted scores are closest to the actual expert ratings.
Handles Complex Chemistry: Wine quality depends on messy, non-linear chemical interactions. Random Forest combines many decision trees to capture these patterns much better than a straight line (Linear Regression) or a single overfitted tree.

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
from sklearn.metrics import roc_auc_score, roc_curve
```

```python
lr_model = LogisticRegression(random_state=42, max_iter=1000)
dt_model = DecisionTreeClassifier(random_state=42)
rf_model = RandomForestClassifier(random_state=42)

# 2. Train (fit) each model on the training data
lr_model.fit(X_train, y_train)
dt_model.fit(X_train, y_train)
rf_model.fit(X_train, y_train)

print("All three classification models have been successfully trained!")
```

```python
y_pred_lr = lr_model.predict(X_test)
y_pred_dt = dt_model.predict(X_test)
y_pred_rf = rf_model.predict(X_test)

predictions_dict = {
    'Logistic Regression': y_pred_lr,
    'Decision Tree': y_pred_dt,
    'Random Forest': y_pred_rf
}

print("Predictions generated successfully for all models!")
```

```python
evaluation_results = []

for name, y_pred in predictions_dict.items():
    # Calculate metrics for the minority class ('Good')
    acc = accuracy_score(y_test, y_pred)
    prec = precision_score(y_test, y_pred, pos_label='Good', zero_division=0)
    rec = recall_score(y_test, y_pred, pos_label='Good', zero_division=0)
    f1 = f1_score(y_test, y_pred, pos_label='Good', zero_division=0)
    
    evaluation_results.append({
        'Model': name,
        'Accuracy': round(acc, 4),
        'Precision (Good)': round(prec, 4),
        'Recall (Good)': round(rec, 4),
        'F1-Score (Good)': round(f1, 4)
    })

# 2. Convert results into a clean comparison DataFrame
comparison_df = pd.DataFrame(evaluation_results)
print("Classification Model Performance Comparison:")
print(comparison_df.to_string(index=False))
```

```python
cm = confusion_matrix(y_test, y_pred_rf, labels=['Good', 'Not Good'])

disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=['Good', 'Not Good'])

fig, ax = plt.subplots(figsize=(6, 6))
disp.plot(cmap=plt.cm.Blues, ax=ax, values_format='d')
plt.title('Confusion Matrix - Random Forest Classifier')
plt.grid(False) # Removes grid lines for cleaner visualization
plt.show()
```

```python
y_test_binary = (y_test == 'Good').astype(int)

y_prob_rf = rf_model.predict_proba(X_test)[:, list(rf_model.classes_).index('Good')]

roc_auc = roc_auc_score(y_test_binary, y_prob_rf)
print(f"Random Forest ROC-AUC Score: {roc_auc:.4f}")

fpr, tpr, thresholds = roc_curve(y_test_binary, y_prob_rf)

plt.figure(figsize=(6, 6))
plt.plot(fpr, tpr, label=f'Random Forest (AUC = {roc_auc:.3f})', color='blue', linewidth=2)
plt.plot([0, 1], [0, 1], 'k--', label='Random Guessing (AUC = 0.500)')
plt.xlabel('False Positive Rate (FPR)')
plt.ylabel('True Positive Rate (TPR)')
plt.title('Receiver Operating Characteristic (ROC) Curve')
plt.legend(loc='lower right')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

Strengths of Random Forest:

1.Captures Non-Linearity: Excels at modeling complex chemical interactions between wine features.
2.Resistant to Overfitting: Uses bagging (averaging multiple trees) to reduce variance and improve test generalization.
3.No Scaling Needed: Invariant to monotonic transformations, so features do not need normalization.
4.Handles Noise: Robust against outliers and measurement errors in the dataset.

Limitations of Random Forest:

1."Black-Box" Nature: Harder to interpret individual decision paths compared to Logistic Regression or single Decision Trees.
2.Computationally Expensive: Requires more memory and processing power to train and store hundreds of trees.
3.Hyperparameter Sensitivity: Performance relies heavily on tuning parameters like tree depth and number of estimators.
