Regesssion model:
Linear Regression: Lowest performance ($R^2 \approx 0.28-0.32$) and highest error. It assumes strict linear relationships, leading to high bias because wine quality depends on complex, non-linear chemical interactions.

Decision Tree Regressor: Moderate performance. It captures non-linear patterns well, but unconstrained trees suffer from high variance and overfit the data.

Random Forest Regressor (Winner): Highest performance ($R^2 \approx 0.45-0.52$) and lowest error metrics (MAE/RMSE). By averaging an ensemble of decision trees (bagging), it captures non-linear interactions while minimizing variance and overfitting.

Classification model:
Logistic Regression: Struggles with class imbalance. It defaults to predicting the majority "Not Good" class, yielding low Recall and F1-scores for high-quality wines due to its linear decision boundary.

Decision Tree: Captures non-linear feature interactions better than linear models, but suffers from instability and high variance, leading to lower Precision on unseen test data.

Random Forest (Best Model): Combines multiple decision trees via bagging. It effectively resolves non-linear chemical interactions and mitigates class imbalance, achieving the top scores across ROC-AUC, F1-Score, and overall accuracy.

Alcohol Content: Consistently ranks as the strongest positive predictor, as higher alcohol levels strongly align with higher perceived white wine quality.

Volatile Acidity: Ranks as a major negative predictor; excessive acetic acid (vinegar taste) sharply degrades quality ratings.

Density & Residual Sugar: Act as important secondary indicators reflecting body, sweetness, and mouthfeel balance.

Free and Total Sulfur Dioxide: Essential for preservation, freshness, and preventing oxidation.

OLS Coefficients (Linear Regression): Measure the linear rate of change in the target variable for a one-unit increase in a feature, holding others constant. They are sensitive to feature scaling and multi-collinearity, and assume a strictly additive, linear relationship.

Random Forest Feature Importance: Measures how much model performance (e.g., ROC-AUC or impurity reduction) drops when a feature's values are shuffled. It does not assume linearity, can capture complex multi-variable interactions, and handles features regardless of scale.

Key Difference: OLS highlights linear directional impact (positive or negative slope), whereas Random Forest importance captures overall predictive utility, including non-linear thresholds and interaction effects.

Statistical Association (Explanatory Modeling): Focuses on inference and hypothesis testing (e.g., "Is there a statistically significant relationship between volatile acidity and wine quality?"). It relies on metrics like p-values, confidence intervals, and coefficients to explain how variables relate within a specific dataset.

Predictive Importance (Machine Learning): Focuses on generalizability and forecasting power (e.g., "How much does knowing alcohol content improve our ability to predict the quality of an unseen bottle of wine?"). It evaluates out-of-sample accuracy rather than population-level statistical significance.

Underfitting: Observed in Linear Regression, which exhibited high bias and low $R^2$ scores ($\approx 0.28-0.32$). It failed to capture the complex, non-linear chemical interactions that determine wine taste.

Overfitting: Unconstrained single Decision Trees showed tendencies to overfit by growing excessively deep, memorizing training set noise, and generalizing poorly to test data (high variance).

Optimal Balance: The Random Forest model successfully mitigated both issues by averaging multiple decorrelated trees via bagging, achieving strong generalization on the test set.

Dataset Limitations:

Subjective Target Variable: Wine quality scores are based on human sensory evaluations, which introduce personal bias and inconsistency.

Class Imbalance: The binary split ("Good" $\ge 7$ vs. "Not Good" $< 7$) results in an imbalanced distribution (~78% vs. ~22%), requiring careful metric selection (Precision, Recall, F1, ROC-AUC) over raw accuracy.

Missing Context: The dataset lacks categorical metadata such as vintage year, grape region, or price point, which heavily influence consumer perception.

Model Limitations:

Random Forest Black-Box Nature: Difficult to trace exact decision pathways for individual predictions compared to linear models.

Computational Cost: Ensemble models require more memory and training time.

Reasons for Recommendation:

Superior Performance: Achieves the highest ROC-AUC score ($\approx 0.85+$) and the strongest balance between Precision and Recall for the minority "Good" class.

Handles Non-Linearity: Accurately models complex, real-world chemical interactions (e.g., how alcohol and acidity interact) where linear models fail.

Robustness: Bagging minimizes overfitting and ensures stable predictions on unseen test data compared to single decision trees or sensitive linear boundaries.
